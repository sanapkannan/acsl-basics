```python
# 1. Output: Printing text and variables to the screen
print("Hello, ACSL Junior!")

# 2. Variables and Data Types
age = 14            # Integer (whole number)
score = 95.5        # Float (decimal number)
team_name = "Alpha" # String (text)
is_coding = True    # Boolean (True or False)

# 3. Input: Reading data from the user (always returns a string)
user_input = input("Enter a number: ")  
# Convert the string input to an integer to do math
number = int(user_input) 
print("You entered:", number)
```

```python
a = 17
b = 5

print(a + b)   # Addition -> 22
print(a - b)   # Subtraction -> 12
print(a * b)   # Multiplication -> 85
print(a / b)   # Standard Division -> 3.4
print(a // b)  # Floor Division (removes decimal) -> 3
print(a % b)   # Modulo (remainder of 17 / 5) -> 2
print(b ** 3)  # Exponentiation (5 cubed) -> 125
```

```python
score = 12

# ACSL problems often require categorizing scores or data
if score >= 90:
    print("Category A")
elif score >= 80:
    print("Category B")
else:
    print("Category C")

# Combining conditions with 'and', 'or', 'not'
age = 14
if age >= 11 and age <= 14:
    print("Eligible for Junior Division")
else:
    print("Not eligible for Junior Division")
```

```python
# For Loop: Repeat a fixed number of times (0 to 4)
for i in range(5):
    print("Count:", i)

# For Loop with custom range: range(start, stop_exclusive, step)
for j in range(2, 11, 2):
    print("Even number:", j)  # Prints 2, 4, 6, 8, 10

# While Loop: Repeat until a condition is no longer true
countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1  # Subtract 1 from countdown
```

```python
text = "ACSL competition"

# 1. Length of a string
print(len(text))  # Output: 16

# 2. String Indexing (0-indexed)
print(text[0])   # First character -> 'A'
print(text[-1])  # Last character -> 'n'

# 3. String Slicing: text[start:stop_exclusive]
print(text[0:4])   # Output: 'ACSL'
print(text[5:])    # From index 5 to end -> 'competition'
print(text[::-1])  # Reverse the entire string -> 'noititepmoc LSCA'

# 4. Built-in String Methods
word = "  Python Code  "
print(word.strip())         # Removes outside spaces -> 'Python Code'
print(word.lower())         # Converts to lowercase -> '  python code  '
print(word.upper())         # Converts to uppercase -> '  PYTHON CODE  '
print(word.replace("o", "X")) # Replaces letters -> '  PythXn CXde  '

```

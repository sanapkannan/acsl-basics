```python
def solve(input_string):

    text = input_string.upper()
    
    vowels = "AEIOU"
    vowel_count = 0
    for char in text:
        if char in vowels:
            vowel_count += 1

    last_char = text[-1]
    
    shifted_ascii = ord(last_char) + 3
    shifted_char = chr(shifted_ascii)
    
    return f"{vowel_count} {shifted_char}"
```

```python
print(solve("python"))  # Should output: 1 Q
```
